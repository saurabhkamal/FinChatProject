```python
import os
from pypdf import PdfWriter

# Folder containing your PDF files
folder = r"C:\Users\sharm\Desktop\DSProjects\genai-agentiai-nlp\FinChatProject\Data"

# Output (merged) file
output_path = os.path.join(folder, "merged_AI_tech_reports.pdf")

# List all .pdf files in the folder
pdf_files = [f for f in os.listdir(folder) if f.lower().endswith(".pdf")]

# Sort if you want them in alphabetic order
pdf_files.sort(key=str.lower)

writer = PdfWriter()

for fname in pdf_files:
    path = os.path.join(folder, fname)
    print("Adding:", path)
    writer.append(path)

# Write out the merged PDF
with open(output_path, "wb") as out_f:
    writer.write(out_f)

print("Merged PDF created at:", output_path)
```
